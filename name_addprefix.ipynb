```python
import random
from tqdm import tqdm

# Titles with weights
titles = {
    "military": [
        {"full": "พลเอก", "abbreviation": "พล.อ.", "weight": 1},
        {"full": "พลโท", "abbreviation": "พล.ท.", "weight": 1},
        {"full": "พลตรี", "abbreviation": "พล.ตร.", "weight": 1},
        {"full": "พันเอก", "abbreviation": "พ.อ.", "weight": 1},
        {"full": "พันโท", "abbreviation": "พ.ท.", "weight": 1},
        {"full": "พันตรี", "abbreviation": "พ.ต.", "weight": 1},
        {"full": "ร้อยเอก", "abbreviation": "ร.อ.", "weight": 1},
        {"full": "ร้อยโท", "abbreviation": "ร.ท.", "weight": 1},
        {"full": "ร้อยตรี", "abbreviation": "ร.ต.", "weight": 1},
        {"full": "จ่าเอก", "abbreviation": "จ.อ.", "weight": 1},
        {"full": "จ่าโท", "abbreviation": "จ.ท.", "weight": 1},
        {"full": "จ่าตรี", "abbreviation": "จ.ต.", "weight": 1},
        {"full": "พลเรือเอก", "abbreviation": "พล.ร.อ.", "weight": 1},
        {"full": "พลเรือโท", "abbreviation": "พล.ร.ท.", "weight": 1},
        {"full": "พลเรือตรี", "abbreviation": "พล.ร.ต.", "weight": 1},
        {"full": "พันเรือเอก", "abbreviation": "พ.ร.อ.", "weight": 1},
        {"full": "พันเรือโท", "abbreviation": "พ.ร.ท.", "weight": 1},
        {"full": "พันเรือตรี", "abbreviation": "พ.ร.ต.", "weight": 1},
        {"full": "ร้อยเรือเอก", "abbreviation": "ร.ร.อ.", "weight": 1},
        {"full": "ร้อยเรือโท", "abbreviation": "ร.ร.ท.", "weight": 1},
        {"full": "ร้อยเรือตรี", "abbreviation": "ร.ร.ต.", "weight": 1},
        {"full": "พลอากาศเอก", "abbreviation": "พล.อ.อ.", "weight": 1},
        {"full": "พลอากาศโท", "abbreviation": "พล.อ.ท.", "weight": 1},
        {"full": "พลอากาศตรี", "abbreviation": "พล.อ.ต.", "weight": 1},
        {"full": "พันอากาศเอก", "abbreviation": "พ.อ.อ.", "weight": 1},
        {"full": "พันอากาศโท", "abbreviation": "พ.อ.ท.", "weight": 1},
        {"full": "พันอากาศตรี", "abbreviation": "พ.อ.ต.", "weight": 1},
        {"full": "ร้อยอากาศเอก", "abbreviation": "ร.อ.อ.", "weight": 1},
        {"full": "ร้อยอากาศโท", "abbreviation": "ร.อ.ท.", "weight": 1},
        {"full": "ร้อยอากาศตรี", "abbreviation": "ร.อ.ต.", "weight": 1}
    ],
    "academic": [
        {"full": "ศาสตราจารย์", "abbreviation": "ศ.", "weight": 4},
        {"full": "รองศาสตราจารย์", "abbreviation": "รศ.", "weight": 3},
        {"full": "ผู้ช่วยศาสตราจารย์", "abbreviation": "ผศ.", "weight": 3},
        {"full": "ดอกเตอร์", "abbreviation": "ดร.", "weight": 2},
    ],
    "normal": [
        {"full": "นาย", "abbreviation": "นาย", "weight": 4},
        {"full": "นาง", "abbreviation": "น.", "weight": 4},
        {"full": "นางสาว", "abbreviation": "น.ส.", "weight": 4},
        {"full": "เด็กชาย", "abbreviation": "ดช.", "weight": 2},
        {"full": "เด็กหญิง", "abbreviation": "ดญ.", "weight": 2},
    ],
    "empty": [
        {"full": "", "abbreviation": "", "weight": 32},
    ],
}

# Function to generate names with random prefixes
def generate_names_with_random_prefixes(names, titles):
    random_names = []
    
    # Create a combined list of titles with weights
    combined_titles = []
    combined_weights = []
    
    for category in titles:
        for title in titles[category]:
            combined_titles.append(title)  # Append the whole title object
            combined_weights.append(title["weight"])
    
    for name in tqdm(names):
        selected_title = random.choices(combined_titles, weights=combined_weights)[0]
        title_variant = random.choice([selected_title["full"], selected_title["abbreviation"]])  # Randomly choose full or abbreviation
        if title_variant == "":
            random_names.append(f"{name}")
        else:
            space = random.random() > 0.5
            if space:
                random_names.append(f"{title_variant} {name}")
            else:
                random_names.append(f"{title_variant}{name}")
    
    return random_names

# Generate names
result = generate_names_with_random_prefixes(names, titles)
result[:20]
```

```python
import pandas as pd

firstname_df = pd.read_csv("/project/lt200324-optmul/pluem/csv_data/firstname_corpus_18sep.csv")
lastname_df = pd.read_csv("/project/lt200324-optmul/pluem/csv_data/lastname_corpus_21aug.csv")
```

```python
from tqdm import tqdm

def generate_random_names(firstname_df, lastname_df, num_combinations=1):
    results = []
    for _ in tqdm(range(num_combinations)):
        # Randomly select a first name and a last name
        first_name = firstname_df.sample(n=1).iloc[0]
        last_name = lastname_df.sample(n=1).iloc[0]

        # Combine the names
        full_name = f"{first_name['firstname']} {last_name['lastname']}"

        results.append({'name': full_name})

    return pd.DataFrame(results)
```

```python
# Generate random combinations
result = generate_random_names(firstname_df, lastname_df, num_combinations=5600000)
result
```

```python
# Remove duplicate rows based on 'name' columns
new_result = result.drop_duplicates(subset=['name'])
new_result = new_result.reset_index(drop=True)
new_result
```
